Restaurant Billing System

```python
# Menu stored as a dictionary: {item_number: (item_name, price)}
menu = {
    1: ("Burger", 120),
    2: ("Pizza", 250),
    3: ("Pasta", 180),
    4: ("French Fries", 90),
    5: ("Cold Coffee", 80),
}


def display_menu():
  """Displays the available food items and their prices."""
  print("\n--- RESTAURANT MENU ---")
  print(f"{'No.':<5} {'Item':<18} {'Price (₹)':<10}")
  print("-" * 35)
  for num, (item, price) in menu.items():
    print(f"{num:<5} {item:<18} ₹{price:<10}")
  print("-" * 35)


def calculate_bill(order):
  """Calculates subtotal, 5% GST tax, and final grand total."""
  subtotal = 0
  for item_num, qty in order.items():
    price = menu[item_num][1]
    subtotal += price * qty

  tax = subtotal * 0.05  # 5% GST
  grand_total = subtotal + tax
  return subtotal, tax, grand_total


def generate_bill(order):
  """Prints the itemized receipt."""
  if not order:
    print("\nNo items ordered. Thank you for visiting!")
    return

  print("\n========== FINAL BILL ==========")
  print(f"{'Item':<16} {'Qty':<5} {'Total (₹)':<10}")
  print("-" * 32)

  for item_num, qty in order.items():
    name, price = menu[item_num]
    item_total = price * qty
    print(f"{name:<16} {qty:<5} ₹{item_total:<10}")

  subtotal, tax, grand_total = calculate_bill(order)
  print("-" * 32)
  print(f"{'Subtotal:':<22} ₹{subtotal:.2f}")
  print(f"{'GST (5%):':<22} ₹{tax:.2f}")
  print(f"{'Grand Total:':<22} ₹{grand_total:.2f}")
  print("================================")
  print("Thank you! Visit again.")


# Main program execution
def main():
  order = {}
  display_menu()

  while True:
    try:
      choice = int(input("\nEnter item number to order (or 0 to finish): "))
      if choice == 0:
        break
      elif choice in menu:
        qty = int(input(f"Enter quantity for {menu[choice][0]}: "))
        if qty > 0:
          order[choice] = order.get(choice, 0) + qty
          print(f"Added {qty} x {menu[choice][0]}.")
        else:
          print("Quantity must be at least 1.")
      else:
        print("Invalid item number. Please choose from the menu.")
    except ValueError:
      print("Invalid input. Please enter numbers only.")

  generate_bill(order)


if __name__ == "__main__":
  main()
```
